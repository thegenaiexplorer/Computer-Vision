## Introduction to Automatic License Plate Recognition

Automatic License Plate Recognition or ALPR for short is a real-world use case. It is the automatic detection and recognition of characters on the number plate of a vehicle.

Starting from highway traffic monitoring, toll control, to recongnizing stolen vehicles, it has a wide range of applications.

As the name suggests, it is a two-stage process:
1. First, we need to detect the license plates on the vehicles.
2. Then we need to provide the cropped images of the detected license plates to an OCR system that does the character recognition.

Over the year, both, the detection pipeline and OCR pipeline have gone through many advancements. This makes building a simple ALPR system comparatively easier today.

We have many state-of-the-art object detection models which can run in real-time even on laptop GPUs. Moreover, transformer based OCR models (e.g. TrOCR that we learned about in the previous sections) can be used out of box with very high accuracy.

This brings us to the approach that we will use in this case-study project.



## The License Plate Detection Model

We will use the latest YOLOv10 model for license plate detection.

In particular we will train the YOLOv10 medium model. We will use license plate images from Google Open Images.

The training dataset consists of 1500 images and the validation dataset consists of 300 images.

## License Plate Recognition

For license plate recognition, we will use the TrOCR stage-1 pretrained models.

The stage-1 pretrained models were not fine-tuned on the handwritten dataset or the printed receipts dataset. They tend to perform better on images from the wild where the context and color can vary from one license plate to another.
