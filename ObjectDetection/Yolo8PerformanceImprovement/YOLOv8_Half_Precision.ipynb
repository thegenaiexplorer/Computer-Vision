# Half Precision

```python
from ultralytics import YOLO
```

```python
yolist = ['yolov8l.pt', 'yolo26l.pt']
```

```python
def load_mAP(model_name, half, imgsz):
  model = YOLO(model_name)
  results = model.val(data="coco128.yaml", imgsz=imgsz, iou=0.5, half=half, verbose=False)
  return results.results_dict
```

```python
result = load_mAP(yolist[0], False, 640)
result
```

```python
result = load_mAP(yolist[0], True, 640)
result
```

```python
result = load_mAP(yolist[1], False, 640)
result
```

```python
result = load_mAP(yolist[1], True, 640)
result
```

As we see from above results we get a speed up of more than 50% when we reduce the model precision to 16bitFP with almost no loss of accuracy.
