### Speed

```python
from ultralytics import YOLO
```

```python
yolist = ['/content/runs/detect/yolov8_sperm-2/weights/best.pt', '/content/runs/detect/yolov8_sperm_small/weights/best.pt']
```

```python
def load_mAP(model_name, half, imgsz=640):
  model = YOLO(model_name)
  results = model.val(data="data/sperm.yaml", iou=0.5, half=half, verbose=False)
  return results.results_dict
```

```python
result = load_mAP(yolist[0], False)
result
```

```python
result = load_mAP(yolist[0], True)
result
```

```python
result = load_mAP(yolist[1], False)
result
```

```python
result = load_mAP(yolist[1], True)
result
```
