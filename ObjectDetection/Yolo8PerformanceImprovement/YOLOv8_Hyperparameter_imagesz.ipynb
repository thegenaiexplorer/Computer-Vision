## Effect of image size on precision and inference speed

```python
yolist = ['yolov8l.pt', 'yolo26l.pt']
```

```python
def load_mAP(model_name, imgsz):
  model = YOLO(model_name)
  results = model.val(data="coco128.yaml", imgsz=imgsz, verbose=False)
  return results.results_dict
```

```python
result = load_mAP(yolist[0], 640)
result
```

```python
result = load_mAP(yolist[1], 640)
result
```

```python
result = load_mAP(yolist[0], 320)
result
```

```python
result = load_mAP(yolist[1], 320)
result
```

As we can see above, mAP goes down when the image size is reduced. Also the inference times decrease when we reduce the image size. This is an expected phenomenon
